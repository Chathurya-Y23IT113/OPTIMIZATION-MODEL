```python
from pulp import *
import pandas as pd
import matplotlib.pyplot as plt
model = LpProblem("Profit_Maximization", LpMaximize)
A = LpVariable('Product_A', lowBound=0)
B = LpVariable('Product_B', lowBound=0)
model += 40 * A + 30 * B
model += 2 * A + B <= 100      
model += A + B <= 80          
model += 3 * A + 2 * B <= 180
model.solve()
print("Status:", LpStatus[model.status])
print("Optimal Units of Product A:", value(A))
print("Optimal Units of Product B:", value(B))
print("Maximum Profit: ₹", value(model.objective))
results = pd.DataFrame({
    'Product': ['Product A', 'Product B'],
    'Optimal Units': [value(A), value(B)]
})
print("\nResults Table:")
print(results)
products = ['Product A', 'Product B']
units = [value(A), value(B)]
plt.figure(figsize=(6,5))
plt.bar(products, units)
plt.xlabel('Products')
plt.ylabel('Optimal Units')
plt.title('Optimal Production Plan')

plt.show()
```
